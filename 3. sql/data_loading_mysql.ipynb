```python
from sqlalchemy import create_engine
from urllib.parse import quote_plus #added this because of special characters in password

username = 'root'
password =  quote_plus('your_password') # Used quote_plus to encode special characters
host = 'localhost'
port = '3306'
database = 'consumer_behaviour'

engine = create_engine(f'mysql+pymysql://{username}:{password}@{host}:{port}/{database}')
table_name = 'consumer_data'

df.to_sql(name=table_name, con=engine, if_exists='replace', index=False) #instead of replace we can choose append also

query = f'SELECT * FROM {table_name} LIMIT 5;'
pd.read_sql(query, engine)
```
